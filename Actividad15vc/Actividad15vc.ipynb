## 1.1 La naturaleza como optimizadora

En esta lectura se explica que la naturaleza puede entenderse como un gran sistema de optimización. A través del tiempo, los seres vivos se han ido adaptando a su entorno mediante pequeños cambios heredables. Aquellos individuos que presentan características favorables para sobrevivir y reproducirse tienen mayores probabilidades de conservar esas características en las siguientes generaciones.

Esta idea está relacionada con la **selección natural**, propuesta por Darwin. La naturaleza no genera seres perfectos de forma inmediata, sino que produce muchos cambios a lo largo del tiempo y conserva los que resultan útiles. De esta manera, las especies se van ajustando a su ambiente de forma gradual.

La lectura también menciona que este comportamiento inspiró a varios científicos a desarrollar métodos computacionales basados en la evolución. Así nacieron enfoques como las estrategias evolutivas, la programación evolutiva y los algoritmos genéticos. Todos ellos intentan imitar el mecanismo natural de mejora: crear varias posibles soluciones, evaluarlas, conservar las mejores y generar nuevas variantes.

En mi opinión, esta parte de la lectura es importante porque muestra que la naturaleza puede verse como un modelo inteligente de resolución de problemas. En lugar de buscar una solución perfecta desde el inicio, se trabaja con múltiples intentos y se mejora poco a poco. Esa misma lógica puede aplicarse en la computación para resolver problemas complejos.

## 1.2 Un poco de biología

En esta sección se presentan algunos conceptos biológicos fundamentales que ayudan a comprender mejor la base de los algoritmos genéticos. Uno de los más importantes es la diferencia entre **genotipo** y **fenotipo**. El genotipo representa la información genética de un individuo, mientras que el fenotipo corresponde a las características observables que resultan de esa información y de la influencia del ambiente.

También se explica que la información genética está contenida en los **cromosomas**, los cuales están formados por ADN. Dentro del ADN se encuentran los **genes**, y las diferentes variantes de un mismo gen se llaman **alelos**. Cada individuo hereda información de sus padres, lo que da lugar a combinaciones únicas de características.

Otro concepto relevante es el de la **meiosis**, proceso mediante el cual se forman los gametos. Durante este proceso ocurre el **cruzamiento** o recombinación, donde se mezclan fragmentos de material genético del padre y de la madre. Esto genera variabilidad en los descendientes. Además, durante la replicación del ADN pueden presentarse errores llamados **mutaciones**, que normalmente son desfavorables, aunque en algunos casos pueden aportar ventajas.

Estos conceptos son importantes en cómputo evolutivo porque sirven como analogía directa. En un algoritmo genético, un individuo puede representarse como una posible solución, sus genes como los valores que lo describen, la población como el conjunto de soluciones, la mutación como un cambio aleatorio y la selección como el proceso de elegir a los mejores individuos.

Considero que esta parte de la lectura ayuda a entender que los algoritmos genéticos no surgieron de manera arbitraria, sino como una abstracción de procesos biológicos reales. Gracias a ello, es posible modelar computacionalmente la idea de evolución para buscar mejores soluciones.

## Relación entre la lectura y el programa desarrollado

A partir de las ideas revisadas en la lectura, se desarrollaron dos clases en Python: `Individuo` y `Poblacion`.

La clase `Individuo` representa una posible solución dentro de un proceso evolutivo. Cada individuo posee genes, un fenotipo y una aptitud. Los genes representan la información básica del individuo; el fenotipo es una forma observable de esa información; y la aptitud indica qué tan buena es la solución que representa.

La clase `Poblacion` agrupa a varios individuos y permite simular una pequeña evolución. Dentro de la población se puede evaluar a los individuos, seleccionar a los mejores, reproducirlos y aplicar mutaciones. Esto refleja la lógica de la selección natural y de los algoritmos genéticos.

Con este programa se busca representar de forma sencilla los conceptos principales del cómputo evolutivo vistos en la lectura, reforzando la conexión entre biología y programación.

```python
import random

class Individuo:
    def __init__(self, longitud_genes=5):
        self.longitud_genes = longitud_genes
        self.genes = [random.randint(0, 1) for _ in range(longitud_genes)]
        self.fenotipo = self.obtener_fenotipo()
        self.aptitud = self.evaluar_aptitud()

    def obtener_fenotipo(self):
        """
        En este ejemplo, el fenotipo será la suma de los genes activos.
        """
        return sum(self.genes)

    def evaluar_aptitud(self):
        """
        La aptitud se calcula como la suma de genes con valor 1.
        Mientras más unos tenga, mejor será la aptitud.
        """
        return sum(self.genes)

    def mutar(self, probabilidad=0.1):
        """
        Cambia algunos genes aleatoriamente con cierta probabilidad.
        """
        for i in range(len(self.genes)):
            if random.random() < probabilidad:
                self.genes[i] = 1 - self.genes[i]

        self.fenotipo = self.obtener_fenotipo()
        self.aptitud = self.evaluar_aptitud()

    def mostrar_info(self):
        return {
            "genes": self.genes,
            "fenotipo": self.fenotipo,
            "aptitud": self.aptitud
        }
```

```python
class Poblacion:
    def __init__(self, tamaño=6, longitud_genes=5):
        self.tamaño = tamaño
        self.longitud_genes = longitud_genes
        self.individuos = [Individuo(longitud_genes) for _ in range(tamaño)]
        self.generacion = 1

    def evaluar_poblacion(self):
        """
        Evalúa nuevamente a cada individuo.
        """
        for individuo in self.individuos:
            individuo.fenotipo = individuo.obtener_fenotipo()
            individuo.aptitud = individuo.evaluar_aptitud()

    def mostrar_poblacion(self):
        """
        Muestra la información de todos los individuos.
        """
        print(f"\n--- Generación {self.generacion} ---")
        for i, individuo in enumerate(self.individuos, start=1):
            print(f"Individuo {i}: {individuo.mostrar_info()}")

    def seleccionar_mejores(self, cantidad=2):
        """
        Selecciona a los individuos con mayor aptitud.
        """
        ordenados = sorted(self.individuos, key=lambda x: x.aptitud, reverse=True)
        return ordenados[:cantidad]

    def reproducir(self, padre1, padre2):
        """
        Genera un hijo combinando genes de dos padres.
        """
        punto_cruce = random.randint(1, self.longitud_genes - 1)

        hijo = Individuo(self.longitud_genes)
        hijo.genes = padre1.genes[:punto_cruce] + padre2.genes[punto_cruce:]
        hijo.fenotipo = hijo.obtener_fenotipo()
        hijo.aptitud = hijo.evaluar_aptitud()

        return hijo

    def evolucionar(self):
        """
        Genera una nueva población a partir de los mejores individuos.
        """
        self.evaluar_poblacion()
        mejores = self.seleccionar_mejores(2)

        nueva_poblacion = []

        # Se conservan los dos mejores
        nueva_poblacion.append(mejores[0])
        nueva_poblacion.append(mejores[1])

        # Se generan nuevos hijos hasta completar la población
        while len(nueva_poblacion) < self.tamaño:
            hijo = self.reproducir(mejores[0], mejores[1])
            hijo.mutar(probabilidad=0.2)
            nueva_poblacion.append(hijo)

        self.individuos = nueva_poblacion
        self.generacion += 1
```

## Prueba de funcionamiento

A continuación se crea una población inicial, se muestra su contenido y luego se simulan algunas generaciones para observar cómo cambia la población con el tiempo.

```python
# Crear población inicial
poblacion = Poblacion(tamaño=6, longitud_genes=5)

# Mostrar generación inicial
poblacion.mostrar_poblacion()
```

```python
# Evolucionar la población durante 3 generaciones
for _ in range(3):
    poblacion.evolucionar()
    poblacion.mostrar_poblacion()
```

## Explicación de resultados

Al ejecutar el programa, se observa que la población está formada por varios individuos, cada uno con sus propios genes, fenotipo y aptitud.

Posteriormente, el método `evolucionar()` selecciona a los mejores individuos de la población actual, los utiliza como base para reproducirse y genera nuevos descendientes. Además, se aplica una pequeña mutación para introducir variabilidad.

Este comportamiento representa de forma simplificada los conceptos revisados en la lectura:
- **selección natural**, al elegir a los mejores individuos,
- **herencia**, al combinar genes de los padres,
- **mutación**, al modificar aleatoriamente algunos genes,
- **evolución**, al generar nuevas generaciones.

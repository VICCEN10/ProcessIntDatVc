```python
from random import random
```

```python
# Función que se quiere minimizar
def funcion(x, y):
    sum1 = x**2 * (4 - 2.1*x**2 + x**4/3.0)
    sum2 = x*y
    sum3 = y**2 * (-4 + 4*y**2)
    return sum1 + sum2 + sum3

# Devuelve un número aleatorio dentro de un rango con
# distribución uniforme (proporcionada por random)
def aleatorio(inf, sup):
    return random()*(sup-inf) + inf
```

```python
# Clase que representa una partícula individual y que facilita
# las operaciones necesarias
class Particula:
    # Algunos atributos de clase (comunes a todas las partículas)
    # Parámetros para actualizar la velocidad
    inercia = 1.4
    cognitiva = 2.0
    social = 2.0

    # Límites del espacio de soluciones
    infx = -2.0
    supx = 2.0
    infy = -1.0
    supy = 1.0

    # Factor de ajuste de la velocidad inicial
    ajusteV = 100.0

    # Crea una partícula dentro de los límites indicados
    def __init__(self):
        self.x = aleatorio(Particula.infx, Particula.supx)
        self.y = aleatorio(Particula.infy, Particula.supy)
        self.vx = aleatorio(Particula.infx/Particula.ajusteV,
                            Particula.supx/Particula.ajusteV)
        self.vy = aleatorio(Particula.infy/Particula.ajusteV,
                            Particula.supy/Particula.ajusteV)
        self.xLoc = self.x
        self.yLoc = self.y
        self.valorLoc = funcion(self.x, self.y)

    # Actualiza la velocidad de la partícula
    def actualizaVelocidad(self, xGlob, yGlob):
        cogX = Particula.cognitiva*random()*(self.xLoc-self.x)
        socX = Particula.social*random()*(xGlob-self.x)
        self.vx = Particula.inercia*self.vx + cogX + socX

        cogY = Particula.cognitiva*random()*(self.yLoc-self.y)
        socY = Particula.social*random()*(yGlob-self.x)
        self.vy = Particula.inercia*self.vy + cogY + socY

    # Actualiza la posición de la partícula
    def actualizaPosicion(self):
        self.x = self.x + self.vx
        self.y = self.y + self.vy

        # Debe mantenerse dentro del espacio de soluciones
        self.x = max(self.x, Particula.infx)
        self.x = min(self.x, Particula.supx)
        self.y = max(self.y, Particula.infy)
        self.y = min(self.y, Particula.supy)

        # Si es inferior a la mejor, la adopta como mejor
        valor = funcion(self.x, self.y)
        if valor < self.valorLoc:
            self.xLoc = self.x
            self.yLoc = self.y
            self.valorLoc = valor
```

```python
# Mueve un enjambre de partículas durante las iteraciones indicadas.
# Devuelve las coordenadas y el valor del mínimo obtenido.
def enjambreParticulas(particulas, iteraciones, reduccionInercia):

    # Registra la mejor posición global y su valor
    mejorParticula = min(particulas, key=lambda p:p.valorLoc)
    xGlob = mejorParticula.xLoc
    yGlob = mejorParticula.yLoc
    valorGlob = mejorParticula.valorLoc

    # Bucle principal de simulación
    for iter in range(iteraciones):
        # Actualiza la velocidad y posición de cada partícula
        for p in particulas:
            p.actualizaVelocidad(xGlob, yGlob)
            p.actualizaPosicion()

        # Hasta que no se han movido todas las partículas no se
        # actualiza el mínimo global, para simular que todas se
        # mueven a la vez
        mejorParticula = min(particulas, key=lambda p:p.valorLoc)
        if mejorParticula.valorLoc < valorGlob:
            xGlob = mejorParticula.xLoc
            yGlob = mejorParticula.yLoc
            valorGlob = mejorParticula.valorLoc

        # Finalmente se reduce la inercia de las partículas
        Particula.inercia *= reduccionInercia

    return (xGlob, yGlob, valorGlob)
```

```python
# Parámetros del problema
nParticulas = 10
iteraciones = 100
redInercia = 0.9

# Genera un conjunto inicial de partículas
particulas = [Particula() for i in range(nParticulas)]

# Ejecuta el algoritmo del enjambre de partículas
print("Resultado del código literal de la página 248:")
print(enjambreParticulas(particulas, iteraciones, redInercia))
```

# VERSION ESTRUCTURADA

```python
from random import random, seed

def funcion_e(x, y):
    sum1 = x**2 * (4 - 2.1*x**2 + x**4/3.0)
    sum2 = x*y
    sum3 = y**2 * (-4 + 4*y**2)
    return sum1 + sum2 + sum3

def aleatorio_e(inf, sup):
    return random()*(sup-inf) + inf

def crear_particula(infx, supx, infy, supy, ajusteV):
    x = aleatorio_e(infx, supx)
    y = aleatorio_e(infy, supy)
    vx = aleatorio_e(infx/ajusteV, supx/ajusteV)
    vy = aleatorio_e(infy/ajusteV, supy/ajusteV)

    return {
        "x": x,
        "y": y,
        "vx": vx,
        "vy": vy,
        "xLoc": x,
        "yLoc": y,
        "valorLoc": funcion_e(x, y)
    }

def actualizar_velocidad(p, xGlob, yGlob, inercia, cognitiva, social):
    cogX = cognitiva * random() * (p["xLoc"] - p["x"])
    socX = social * random() * (xGlob - p["x"])
    p["vx"] = inercia * p["vx"] + cogX + socX

    cogY = cognitiva * random() * (p["yLoc"] - p["y"])
    socY = social * random() * (yGlob - p["y"])
    p["vy"] = inercia * p["vy"] + cogY + socY

def actualizar_posicion(p, infx, supx, infy, supy):
    p["x"] = p["x"] + p["vx"]
    p["y"] = p["y"] + p["vy"]

    p["x"] = max(p["x"], infx)
    p["x"] = min(p["x"], supx)
    p["y"] = max(p["y"], infy)
    p["y"] = min(p["y"], supy)

    valor = funcion_e(p["x"], p["y"])
    if valor < p["valorLoc"]:
        p["xLoc"] = p["x"]
        p["yLoc"] = p["y"]
        p["valorLoc"] = valor

def enjambre_estructurado(nParticulas, iteraciones, redInercia, semilla=123):
    seed(semilla)

    inercia = 1.4
    cognitiva = 2.0
    social = 2.0

    infx = -2.0
    supx = 2.0
    infy = -1.0
    supy = 1.0
    ajusteV = 100.0

    particulas = []
    for i in range(nParticulas):
        particulas.append(crear_particula(infx, supx, infy, supy, ajusteV))

    mejor = min(particulas, key=lambda p: p["valorLoc"])
    xGlob = mejor["xLoc"]
    yGlob = mejor["yLoc"]
    valorGlob = mejor["valorLoc"]

    for i in range(iteraciones):
        for p in particulas:
            actualizar_velocidad(p, xGlob, yGlob, inercia, cognitiva, social)
            actualizar_posicion(p, infx, supx, infy, supy)

        mejor = min(particulas, key=lambda p: p["valorLoc"])
        if mejor["valorLoc"] < valorGlob:
            xGlob = mejor["xLoc"]
            yGlob = mejor["yLoc"]
            valorGlob = mejor["valorLoc"]

        inercia *= redInercia

    return (xGlob, yGlob, valorGlob)
```

```python
print("Resultado de la versión estructurada:")
print(enjambre_estructurado(10, 100, 0.9))
```

En este ejercicio se trabajó con el algoritmo de **enjambre de partículas (PSO)** mostrado en la **página 248** y después se realizó una versión usando **programación estructurada**.

La primera implementación conserva la lógica original del código mostrado en el material, utilizando una **clase** para representar cada partícula y sus métodos para actualizar velocidad y posición.  
La segunda implementación realiza el mismo procedimiento, pero ahora mediante **funciones** y estructuras de datos simples, sin utilizar programación orientada a objetos.

Ambas versiones tienen el mismo objetivo: **minimizar la función dada** dentro de los límites establecidos para \(x\) y \(y\).  
También en ambos casos se emplean los mismos elementos principales del algoritmo PSO:

- inicialización de partículas,
- mejor posición local,
- mejor posición global,
- actualización de velocidad,
- actualización de posición,
- reducción de la inercia.

Por lo tanto, la diferencia principal entre ambas implementaciones **no está en el algoritmo**, sino en la **forma de programarlo**.  
La versión del libro está planteada con clases, mientras que la versión estructurada organiza el problema únicamente con funciones y variables.

Al ejecutar ambas versiones, se observa que las dos siguen la misma lógica general de búsqueda del mínimo, por lo que la implementación estructurada representa correctamente el algoritmo presentado en la página 248.

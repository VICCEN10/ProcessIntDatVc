```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
habitaciones = np.array([23,47,21,39,37,29,23,44,45,16,30,42,54,
                         27,34,15,19,38,44,47,43,38,51,61,39], dtype=float)

ingreso = np.array([1452,1361,1426,1470,1456,1430,1354,1442,1394,1459,1399,1458,1537,
                    1425,1445,1439,1348,1450,1431,1446,1485,1405,1461,1490,1426], dtype=float)

df = pd.DataFrame({
    "Dia": np.arange(1, 26),
    "Habitaciones_ocupadas": habitaciones,
    "Ingreso_restaurante": ingreso
})

print(df.head())
```

```python
r_numpy = np.corrcoef(habitaciones, ingreso)[0, 1]
```

```python
n = len(habitaciones)
sum_x = habitaciones.sum()
sum_y = ingreso.sum()
sum_x2 = (habitaciones**2).sum()
sum_y2 = (ingreso**2).sum()
sum_xy = (habitaciones*ingreso).sum()

r_formula = (n*sum_xy - sum_x*sum_y) / np.sqrt((n*sum_x2 - sum_x**2) * (n*sum_y2 - sum_y**2))

print("\nCoeficiente de correlación (numpy):   r =", round(r_numpy, 4))
print("Coeficiente de correlación (fórmula): r =", round(r_formula, 4))
```

```python
m, b = np.polyfit(habitaciones, ingreso, 1)
print("\nRecta de tendencia: Ingreso ≈", round(m, 4), "* Habitaciones +", round(b, 2))
```

```python
plt.figure(figsize=(8, 5.5))
plt.scatter(habitaciones, ingreso)
x_line = np.linspace(habitaciones.min(), habitaciones.max(), 200)
plt.plot(x_line, m*x_line + b)
plt.title("Ingreso del restaurante vs Habitaciones ocupadas (n=25)")
plt.xlabel("Habitaciones ocupadas")
plt.ylabel("Ingreso del restaurante ($)")
plt.grid(True, alpha=0.3)
plt.show()
```
